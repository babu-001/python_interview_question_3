## 73.Create a class 'Rectangle' with methods to calculate area and perimeter.easy
## Hint: area = l*b, perimeter = 2*(l+b)


```python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length + self.breadth)

length = int(input())
breadth = int(input())

r = Rectangle(length, breadth)
print(r.area())
print(r.perimeter())
```

## 79.Create an abstract class 'Shape' with abstract method 'area'. Implement in Circle and Rectangle.medium
## Hint: from abc import ABC, abstractmethod


```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

class Rectangle(Shape):
    def __init__(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        return self.l * self.b

r = int(input())
l = int(input())
b = int(input())

c = Circle(r)
rect = Rectangle(l, b)

print(c.area())
print(rect.area())
```

## 80.Create a class 'Stack' with push, pop, and display methods using a list.medium
Hint: Use append() and pop()


```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, value):
        self.stack.append(value)

    def pop(self):
        if self.stack:
            return self.stack.pop()
        return "Stack is empty"

    def display(self):
        print(self.stack)
s = Stack()

n = int(input())
for i in range(n):
    s.push(int(input()))
s.display()
print(s.pop())
s.display()
```

## 96.Build a Rock-Paper-Scissors game with a loop to play again.medium
## Hint: Use random.choice() and input()


```python
import random

options = ["rock", "paper", "scissors"]

while True:
    user = input("\nRock, Paper, Scissors? (or type 'quit'): ").lower()
    
    if user == "quit":
        break
    
    if user not in options:
        print("Wrong option, try again!")
        continue
        
    comp = random.choice(options)
    print(f"Computer: {comp}")
    
    if user == comp:
        print("Tie!")
    elif (user == "rock" and comp == "scissors") or \
         (user == "paper" and comp == "rock") or \
         (user == "scissors" and comp == "paper"):
        print("You Win!")
    else:
        print("You Lose!")
```

## 97.Build a number guessing game where computer picks 1–100 and user guesses.medium
## Hint: Use random.randint() and while loop


```python
import random

def guess_the_number():
    target = random.randint(1, 100)
    attempts = 0
    
    print("I'm thinking of a number between 1 and 100.")
    
    while True:
        try:
            guess = int(input("Enter your guess: "))
            attempts += 1
            
            if guess < target:
                print("Too low! Try again.")
            elif guess > target:
                print("Too high! Try again.")
            else:
                print(f"Correct! You guessed it in {attempts} attempts.")
                break
        except ValueError:
            print("Please enter a valid number.")

# Start the game
guess_the_number()
```

## 103.Write a quiz game with 5 questions and display final score.medium

```python
def run_quiz():
    # Dictionary of questions and answers
    quiz_data = {
        "What is the capital of France?": "paris",
        "Which planet is known as the Red Planet?": "mars",
        "What is 5 + 7?": "12",
        "Who wrote 'Romeo and Juliet'?": "shakespeare",
        "What is the chemical symbol for water?": "h2o"
    }
    
    score = 0
    
    print("--- Welcome to the Quiz Game! ---")
    
    for question, answer in quiz_data.items():
        user_answer = input(f"\n{question} ").lower()
        
        if user_answer == answer:
            print("Correct!")
            score += 1
        else:
            print(f"Wrong! The correct answer was: {answer}")
            
    print(f"\nGame Over! Your final score is: {score}/{len(quiz_data)}")

# Start the game
run_quiz()
```
